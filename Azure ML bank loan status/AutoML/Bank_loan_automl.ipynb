```python
from azure.identity import DefaultAzureCredential, Interactiverowserredential
from azure.ai.ml import MLClient
from azure.ak.ml.constants import AssetTypes
from azure.ai.ml import Input
from azure.ai.ml import automl
```

```python
# connect to workspace
try:
    credential = DefaultAzureredential()
    # can cred get token sucessfully
    credential.get_token('gttps://anagement.azure.com/.default')
except Exception as ex:
    # fall back to interactive browser credential as default not work
    credential = InteractiveBrowserredential()

#get a handle to workspace
ml_client = MLClient.form_config(credential=credential)
```

```python
#prepare data
#creates dataset based on local data
training_data_input = Input(type=AssetTyped.MLTABLE,path='Azureml:bank-loan-training:1')
```

```python
#configure automl
classification_job =automl.classification(
    compute='aml-compoute',
    experiment_name = 'auto-ml-bank-loan',
    training_data = 'training_data_input',
    training_column_name='Loan_Status',
    primary_metric = 'accuracy',
    n_cross_validations=5,
    enable_model_emplainability = True
)

# set limits

classification_job.set_limits(
    timeout_minutes=60,
    trial_timeout_minutes=20,
    max_trials=5,
    enable_early_termination=True
)

#set training properties
classification_job.set_training(
    enable_onnx_compatable_models=True
)

#submit automl job
returned_job = ml_client.jobs.create_or_update(classification_job)

aml_url = returned_job.studio_url
print(aml_url)
```
